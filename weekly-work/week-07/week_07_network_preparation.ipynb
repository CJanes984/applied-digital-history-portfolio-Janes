# Week 7: Historical Network Analysis

This notebook transforms a synthetic catalogue of letters into node and edge
tables.

```python
import pandas as pd

# 1. Load the dataset into a DataFrame named letters
letters = pd.read_csv('synthetic_correspondence_records.csv')

# 1. Number of rows and columns
print("--- 1. Dimensions ---")
num_rows, num_cols = letters.shape
print(f"Total rows: {num_rows}, Total columns: {num_cols}\n")

# 2. Complete list of column names
print("--- 2. Column Names ---")
print(list(letters.columns), "\n")

# 3. First five rows
print("--- 3. First Five Rows ---")
display(letters.head())

# 4. Whether letter_id is unique
print("\n--- 4. Uniqueness of letter_id ---")
is_unique = letters['letter_id'].is_unique if 'letter_id' in letters.columns else "Column not found"
print(f"Is 'letter_id' strictly unique? {is_unique}\n")

# 5. Number of distinct sender_id values
print("--- 5. Distinct Senders ---")
num_senders = letters['sender_id'].nunique() if 'sender_id' in letters.columns else "Column not found"
print(f"Number of unique senders: {num_senders}\n")

# 6. Number of distinct recipient_id values
print("--- 6. Distinct Recipients ---")
num_recipients = letters['recipient_id'].nunique() if 'recipient_id' in letters.columns else "Column not found"
print(f"Number of unique recipients: {num_recipients}\n")

# 7. Number of distinct people appearing as either sender or recipient
print("--- 7. Total Unique People (Senders & Recipients) ---")
if 'sender_id' in letters.columns and 'recipient_id' in letters.columns:
    unique_people = pd.concat([letters['sender_id'], letters['recipient_id']]).dropna().nunique()
    print(f"Total distinct individuals involved: {unique_people}\n")
else:
    print("Sender or recipient columns missing.\n")

# 8. Counts for date_certainty, document_form, survival_status, and repository
print("--- 8. Value Counts for Key Categorical Fields ---")
target_cols = ['date_certainty', 'document_form', 'survival_status', 'repository']
for col in target_cols:
    if col in letters.columns:
        print(f"\nField: {col}")
        print(letters[col].value_counts(dropna=False))
    else:
        print(f"\nField '{col}' not found.")

# 9. Missing-value counts for every column
print("\n--- 9. Missing-Value Counts per Column ---")
print(letters.isnull().sum(), "\n")

# 10. All records involving an identifier that begins with U rather than P
print("--- 10. Records with 'U' Identifiers (Sender or Recipient) ---")
u_records = letters[
    (letters['sender_id'].astype(str).str.startswith('U', na=False)) |
    (letters['recipient_id'].astype(str).str.startswith('U', na=False))
]
print(f"Found {len(u_records)} records involving a 'U' identifier:")
display(u_records)
```

```python
import pandas as pd

# 11. Do not modify the original letters dataframe
df = letters.copy()

# 1. Create temporary tables for senders and recipients using exact column names
sender_cols = {
    'sender_id': 'Id',
    'sender_name': 'Label',
    'sender_gender': 'gender',
    'sender_role': 'role',
    'sender_city': 'city',
    'sender_confession': 'confession',
    'sender_latitude': 'latitude',
    'sender_longitude': 'longitude'
}

recipient_cols = {
    'recipient_id': 'Id',
    'recipient_name': 'Label',
    'recipient_gender': 'gender',
    'recipient_role': 'role',
    'recipient_city': 'city',
    'recipient_confession': 'confession',
    'recipient_latitude': 'latitude',
    'recipient_longitude': 'longitude'
}

# 2. Extract and rename columns so both tables share identical headers
senders_df = df[list(sender_cols.keys())].rename(columns=sender_cols)
recipients_df = df[list(recipient_cols.keys())].rename(columns=recipient_cols)

# 3. Stack the two temporary tables together
stacked_nodes = pd.concat([senders_df, recipients_df], ignore_index=True)

# 4. Check for conflicting nonblank values across duplicate appearances for each Id
print("--- 4. Checking for Attribute Conflicts per Id ---")
attributes_to_check = ['Label', 'gender', 'role', 'city', 'confession', 'latitude', 'longitude']
conflicts_found = False

for attr in attributes_to_check:
    if attr in stacked_nodes.columns:
        grouped = stacked_nodes.groupby('Id')[attr].apply(lambda x: x.dropna().loc[x.dropna() != ''].unique())
        conflicting_ids = grouped[grouped.apply(len) > 1]

        if not conflicting_ids.empty:
            conflicts_found = True
            print(f"Conflicts found in column '{attr}':")
            for uid, vals in conflicting_ids.items():
                print(f"  Id '{uid}' has conflicting values: {list(vals)}")

if not conflicts_found:
    print("No conflicting nonblank attribute values found.\n")

# 5. Remove duplicate appearances while preserving the unidentified correspondent and missing attributes
nodes = stacked_nodes.drop_duplicates(subset=['Id']).copy()

# 6. Sort the finished table by Id
nodes = nodes.sort_values(by='Id').reset_index(drop=True)

# 7. Confirm that Id is unique and contains no missing values
print("--- 7. Id Validation ---")
id_unique = nodes['Id'].is_unique
id_nulls = nodes['Id'].isnull().sum()
print(f"Is 'Id' unique? {id_unique}")
print(f"Missing values in 'Id': {id_nulls}\n")

# 8. Print the number of nodes and missing-value counts for every column
print("--- 8. Node Summary & Missing Values ---")
print(f"Total number of unique nodes: {len(nodes)}")
print("Missing values per column:")
print(nodes.isnull().sum(), "\n")

# 9. Display the full row for the unidentified correspondent and five other rows
print("--- 9. Sample Rows (Unidentified Correspondent & 5 Others) ---")
unidentified_rows = nodes[nodes['Id'].astype(str).str.startswith('U', na=False)]
if not unidentified_rows.empty:
    print("Unidentified correspondent row(s):")
    display(unidentified_rows)
else:
    print("No ID starting with 'U' found.")

print("Five other sample rows:")
display(nodes.head(5))

# 10. Save the result as "correspondence_nodes.csv" without a pandas index
nodes.to_csv("correspondence_nodes.csv", index=False)
print("\nSuccessfully saved 'correspondence_nodes.csv' without index.")
```

```python
import pandas as pd

# 16. Do not modify letters or nodes
df = letters.copy()

# 1-4. Group records by sender_id and recipient_id, aggregate counts and dates
edges = df.groupby(['sender_id', 'recipient_id'], sort=False).agg(
    Weight=('letter_id', 'count'),
    first_date=('letter_date', 'min'),
    last_date=('letter_date', 'max'),
    surviving_text_count=('survival_status', lambda x: (x == 'text_surviving').sum()),
    catalogue_reference_only_count=('survival_status', lambda x: (x == 'text_not_surviving').sum())
).reset_index()

# 2. Rename sender_id to Source and recipient_id to Target
edges = edges.rename(columns={'sender_id': 'Source', 'recipient_id': 'Target'})

# 3. Add Type with the value "Directed"
edges['Type'] = 'Directed'

# Reorder columns for a clean network structure
edge_cols = [
    'Source', 'Target', 'Type', 'Weight',
    'first_date', 'last_date',
    'surviving_text_count', 'catalogue_reference_only_count'
]
edges = edges[edge_cols]

# 8. Sort result by Weight largest to smallest, then by Source and Target
edges = edges.sort_values(
    by=['Weight', 'Source', 'Target'],
    ascending=[False, True, True]
).reset_index(drop=True)

# 9. Confirm that Source and Target contain no missing values
assert edges['Source'].isnull().sum() == 0, "Source contains missing values!"
assert edges['Target'].isnull().sum() == 0, "Target contains missing values!"

# 10. Confirm that every Source and Target value exists in nodes.Id
unmatched_sources = set(edges['Source']) - set(nodes['Id'])
unmatched_targets = set(edges['Target']) - set(nodes['Id'])
if unmatched_sources or unmatched_targets:
    print(f"Warning - Unmatched Sources: {unmatched_sources}")
    print(f"Warning - Unmatched Targets: {unmatched_targets}")
else:
    print("Validation check: All Source and Target identifiers exist in nodes.Id.")

# 11. Check for self-loops and duplicate Source–Target pairs
self_loops = edges[edges['Source'] == edges['Target']]
duplicates = edges[edges.duplicated(subset=['Source', 'Target'], keep=False)]
print(f"Self-loops found: {len(self_loops)}")
print(f"Duplicate Source-Target pairs found: {len(duplicates)}")

# 12 & 13. Assertions for structural integrity
assert edges['Weight'].sum() == len(df), "Weight sum does not match total letter count!"
assert (edges['surviving_text_count'] + edges['catalogue_reference_only_count'] == edges['Weight']).all(), \
    "Survival counts do not equal Weight for every edge!"

# 14. Print summary metrics and the first ten edges
print("\n--- Summary Statistics ---")
print(f"Input letter count: {len(df)}")
print(f"Output edge count: {len(edges)}")
print(f"Total edge weight: {edges['Weight'].sum()}")
print(f"Minimum Weight: {edges['Weight'].min()}")
print(f"Maximum Weight: {edges['Weight'].max()}")

print("\n--- First Ten Edges ---")
display(edges.head(10))

# 15. Save the result as "correspondence_edges.csv" without a pandas index
edges.to_csv("correspondence_edges.csv", index=False)
print("\nSuccessfully saved 'correspondence_edges.csv' without index.")
```

```python
import networkx as nx
import pandas as pd

# 1. Create a networkx.DiGraph because the edges are directed
G = nx.DiGraph()

# 2 & 3. Add every row in nodes as a graph node, omitting missing attribute values
for _, row in nodes.iterrows():
    node_id = row['Id']
    node_attrs = {}
    for col in ['Label', 'gender', 'role', 'city', 'confession', 'latitude', 'longitude']:
        val = row[col]
        # Only add attribute if it is not missing (omits NaN / None)
        if pd.notna(val):
            node_attrs[col] = val
    G.add_node(node_id, **node_attrs)

# 4, 5 & 6. Add every row in edges as a directed graph edge with weight and attributes
for _, row in edges.iterrows():
    u = row['Source']
    v = row['Target']
    edge_attrs = {
        'weight': row['Weight'],
        'Type': row['Type'],
        'first_date': row['first_date'],
        'last_date': row['last_date'],
        'surviving_text_count': row['surviving_text_count'],
        'catalogue_reference_only_count': row['catalogue_reference_only_count']
    }
    G.add_edge(u, v, **edge_attrs)

# 7. Print graph properties and summary statistics
print("--- 7. Graph Summary ---")
print(f"Is graph directed? {G.is_directed()}")
print(f"Number of nodes: {G.number_of_nodes()}")
print(f"Number of edges: {G.number_of_edges()}")
total_weight = sum(d.get('weight', 0) for u, v, d in G.edges(data=True))
print(f"Sum of all edge weights: {total_weight}\n")

# 8. Confirm that the graph contains exactly the same node identifiers as nodes.Id
assert set(G.nodes()) == set(nodes['Id']), "Graph node identifiers do not match nodes.Id!"
print("Validation 8 passed: All graph node IDs match nodes['Id'].")

# 9. Confirm that every edge and weight in edges appears in G
for _, row in edges.iterrows():
    u, v, w = row['Source'], row['Target'], row['Weight']
    assert G.has_edge(u, v), f"Missing edge in G: Source '{u}' -> Target '{v}'"
    assert G[u][v]['weight'] == w, f"Weight mismatch for edge '{u}' -> '{v}'"
print("Validation 9 passed: All edges and weights in `edges` are correctly represented in G.")

# 10. Write the graph as "correspondence_network.gexf"
nx.write_gexf(G, "correspondence_network.gexf")
print("\nSuccessfully saved graph as 'correspondence_network.gexf'.")

# 11. Read the saved GEXF back into G_check and repeat checks
G_check = nx.read_gexf("correspondence_network.gexf")

print("\n--- 11. G_check Verification ---")
print(f"G_check is directed? {G_check.is_directed()}")
print(f"G_check number of nodes: {G_check.number_of_nodes()}")
print(f"G_check number of edges: {G_check.number_of_edges()}")
check_total_weight = sum(float(d.get('weight', 0)) for u, v, d in G_check.edges(data=True))
print(f"G_check sum of edge weights: {check_total_weight}")

# Assertions for G_check
assert G_check.is_directed() == G.is_directed(), "G_check direction mismatch!"
assert set(G_check.nodes()) == set(G.nodes()), "G_check nodes mismatch!"
assert G_check.number_of_edges() == G.number_of_edges(), "G_check edge count mismatch!"
print("Validation 11 passed: GEXF file successfully read back and verified.")
```

```python
import networkx as nx
import pandas as pd

# 1. Create an unweighted undirected projection from the existing directed graph G
# This records whether two people are connected by at least one catalogued letter
# in either direction, ignoring arrow direction and letter frequency.
G_undirected = G.to_undirected()

# 1. Report the number of nodes and undirected edges
num_nodes = G_undirected.number_of_nodes()
num_edges = G_undirected.number_of_edges()

# 2. Calculate undirected density
density = nx.density(G_undirected)

# 3. Calculate average degree
degrees = [d for n, d in G_undirected.degree()]
avg_degree = sum(degrees) / num_nodes if num_nodes > 0 else 0

# 4. Report the number and sizes of connected components
components = list(nx.connected_components(G_undirected))
num_components = len(components)
component_sizes = [len(c) for c in components]

# 5. Test whether G_undirected is connected
is_connected = nx.is_connected(G_undirected)

# 6 & 7. Calculate path measures (shortest-path length and diameter)
if is_connected:
    avg_path_length = nx.average_shortest_path_length(G_undirected)
    diameter = nx.diameter(G_undirected)
    excluded_nodes = 0
else:
    # If disconnected, calculate only for the largest connected component and report excluded nodes
    largest_cc = max(components, key=len)
    subgraph = G_undirected.subgraph(largest_cc)
    avg_path_length = nx.average_shortest_path_length(subgraph)
    diameter = nx.diameter(subgraph)
    excluded_nodes = num_nodes - len(largest_cc)

# Print results with clear labels
print("--- Undirected Projection Analysis ---")
print(f"1. Number of nodes: {num_nodes}")
print(f"   Number of undirected edges: {num_edges}")
print(f"2. Undirected network density: {density:.4f}")
print(f"3. Average degree: {avg_degree:.4f}")
print(f"4. Connected components: {num_components} component(s) with sizes: {component_sizes}")
print(f"5. Is graph connected? {is_connected}")
if is_connected:
    print(f"6. Average shortest-path length: {avg_path_length:.4f}")
    print(f"   Network diameter: {diameter}")
else:
    print(f"7. Average shortest-path length (Largest Component): {avg_path_length:.4f}")
    print(f"   Network diameter (Largest Component): {diameter}")
    print(f"   Number of excluded nodes (isolated/smaller components): {excluded_nodes}")

# 8 & 9. Store results in a one-row dataframe named network_summary with the model column
summary_data = {
    'model': ['unweighted undirected projection'],
    'num_nodes': [num_nodes],
    'num_edges': [num_edges],
    'density': [density],
    'avg_degree': [avg_degree],
    'num_components': [num_components],
    'is_connected': [is_connected],
    'avg_shortest_path_length': [avg_path_length],
    'diameter': [diameter],
    'excluded_nodes': [excluded_nodes]
}

network_summary = pd.DataFrame(summary_data)

# 10. Save as "correspondence_network_summary.csv" without a pandas index
network_summary.to_csv("correspondence_network_summary.csv", index=False)
print("\nSuccessfully saved summary as 'correspondence_network_summary.csv'.")
```

```python
import networkx as nx
import pandas as pd

# 1. Confirm directedness of both graphs
print("--- 1. Graph Directedness Confirmation ---")
print(f"Is G directed? {G.is_directed()}")
print(f"Is G_undirected directed? {G_undirected.is_directed()}\n")

# 2. Report node count, directed edge count, and total correspondence weight in G
num_nodes = G.number_of_nodes()
num_directed_edges = G.number_of_edges()
total_weight = sum(d.get('weight', 0) for u, v, d in G.edges(data=True))

print("--- 2. Scale & Weight ---")
print(f"Node count: {num_nodes}")
print(f"Directed edge count: {num_directed_edges}")
print(f"Total correspondence weight (total letters): {total_weight}\n")

# 3. Calculate directed density
directed_density = nx.density(G)
print(f"3. Directed density: {directed_density:.4f}")

# 4. Calculate average in-degree, average out-degree, and average total degree
in_degrees = dict(G.in_degree())
out_degrees = dict(G.out_degree())
total_degrees = dict(G.degree())

avg_in_degree = sum(in_degrees.values()) / num_nodes if num_nodes > 0 else 0
avg_out_degree = sum(out_degrees.values()) / num_nodes if num_nodes > 0 else 0
avg_total_degree = sum(total_degrees.values()) / num_nodes if num_nodes > 0 else 0

print(f"4. Average in-degree: {avg_in_degree:.4f}")
print(   f"   Average out-degree: {avg_out_degree:.4f}")
print(   f"   Average total degree: {avg_total_degree:.4f}")

# 5. Calculate reciprocity
reciprocity = nx.reciprocity(G)
print(f"5. Network reciprocity: {reciprocity:.4f}")

# 6. Report the number and sizes of weakly connected components
weak_comps = list(nx.weakly_connected_components(G))
weak_sizes = [len(c) for c in weak_comps]
print(f"6. Weakly connected components: {len(weak_comps)} component(s) with sizes {weak_sizes}")

# 7. Report the number and sizes of strongly connected components
strong_comps = list(nx.strongly_connected_components(G))
strong_sizes = [len(c) for c in strong_comps]
print(f"7. Strongly connected components: {len(strong_comps)} component(s) with sizes {strong_sizes}\n")

# 8 & 9. Calculate in-degree and out-degree centralities and add to node_metrics
in_deg_cent = nx.in_degree_centrality(G)
out_deg_cent = nx.out_degree_centrality(G)

node_metrics = pd.read_csv('correspondence_node_metrics.csv')
node_metrics['in_degree_centrality'] = node_metrics['Id'].map(in_deg_cent)
node_metrics['out_degree_centrality'] = node_metrics['Id'].map(out_deg_cent)
node_metrics.to_csv("correspondence_node_metrics.csv", index=False)

# 10. Display the five highest-scoring nodes for each directed centrality
print("--- 10. Top 5 Nodes by Directed Centrality ---")
print("Top 5 In-Degree Centrality (Receiving Focus):")
display(node_metrics.nlargest(5, 'in_degree_centrality')[['Id', 'Label', 'in_degree_centrality']])

print("\nTop 5 Out-Degree Centrality (Sending Focus):")
display(node_metrics.nlargest(5, 'out_degree_centrality')[['Id', 'Label', 'out_degree_centrality']])

# 12. Add clearly named directed measures to the one-row network_summary dataframe
network_summary = pd.read_csv('correspondence_network_summary.csv')
network_summary['num_directed_edges'] = num_directed_edges
network_summary['total_weight'] = total_weight
network_summary['directed_density'] = directed_density
network_summary['avg_in_degree'] = avg_in_degree
network_summary['avg_out_degree'] = avg_out_degree
network_summary['reciprocity'] = reciprocity
network_summary['num_weakly_connected_components'] = len(weak_comps)
network_summary['num_strongly_connected_components'] = len(strong_comps)

# 13. Save updated dataframes without pandas indexes
network_summary.to_csv("correspondence_network_summary.csv", index=False)
print("\nSuccessfully updated and saved 'correspondence_node_metrics.csv' and 'correspondence_network_summary.csv'.")

# 14. Print several reciprocal pairs and several one-direction-only pairs
reciprocal_pairs = []
one_way_pairs = []
seen = set()

for u, v in G.edges():
    if (v, u) in G:
        if (v, u) not in seen:
            reciprocal_pairs.append((u, v))
            seen.add((u, v))
            seen.add((v, u))
    else:
        one_way_pairs.append((u, v))

print("\n--- 14. Sample Communication Pairs ---")
print(f"Total reciprocal pairs found: {len(reciprocal_pairs)}")
print(f"Sample reciprocal pairs (A <-> B): {reciprocal_pairs[:3]}")
print(f"Total one-direction-only pairs found: {len(one_way_pairs)}")
print(f"Sample one-direction pairs (A -> B): {one_way_pairs[:3]}")
```
